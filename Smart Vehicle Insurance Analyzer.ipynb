```python
# Input data kendaraan dan pengemudi

usia_pengemudi = int(input("Masukkan usia pengemudi (tahun): "))
jenis_kendaraan = input("Masukkan tipe kendaraan (mobil/motor): ").strip().lower()
nilai_kendaraan = float(input("Masukkan nilai kendaraan (dalam rupiah): ")) # Variabel nilai kendaraan
area_penggunaan = input("Masukkan area penggunaan (perkotaan/pedesaan): ").strip().lower()
sistem_pengaman = input("Apakah kendaraan memiliki sistem pengaman? (y/t): ").strip().lower()
```

```python
# 4. Analisis berdasarkan sistem pengaman

if sistem_pengaman == "tidak":
    skor_risiko = 1 # Risiko meningkat satu level 

if skor_risiko > 3:
    skor_risiko = 3
```

```python
# Penentuan kategori risiko dan persentase premi

if skor_risiko == 3:
    kategori_risiko = "Tinggi"
    persentase_premi = 0.1
elif skor_risiko == 2:
    kategori_risiko = "Sedang"
    persentase_premi = 0.07
else:
    kategori_risiko = "Rendah"
    persentase_premi = 0.05
```
