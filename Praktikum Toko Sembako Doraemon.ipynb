```python
# 1. Membuat Data Belanjaan
# Buat list belanjaan
belanjaan = ["Apel", "Susu", "Roti"]
# Buat tuple harga_jumlah (harga1, jumlah1, harga2, jumlah2, harga3, jumlah3)
harga_jumlah = (5000, 3, 12000, 2, 8000, 1)
# Tampilkan karakter pertama dari barang pertama "Apel" dan karakter terakhir dari barang terakhir "Roti"
print(f"Karakter pertama dari barang pertama: {belanjaan[0][0]}")
print(f"Karakter terakhir dari barang terakhir: {belanjaan[-1][-1]}")
```

```python
# 2. Hitung Total Harga per Barang
# Menghitung total harga masing-masing barang secara manual dari tuple
total_apel = harga_jumlah[0] * harga_jumlah[1] # 5000 * 3 = 15000
total_susu = harga_jumlah[2] * harga_jumlah[3] # 12000 * 2 = 24000
total_roti = harga_jumlah[4] * harga_jumlah[5] # 8000 * 1 = 8000
# Menghitung total keseluruhan belanja
total_belanja = total_apel + total_susu + total_roti # 47000
```

```python
# 3. Cetak Struk Belanja
# Format struk menggunakan f-string dan escape sequence (\n, \t)
nama_pembeli = "Nobita Nobi"
nama_toko = "Toko Sembako Doraemon"
# Cetak isi struk
print('======================================')
print('        Toko Sembako Doraemon        ')
print('======================================')
print(f"Pembeli : {nama_pembeli}")
print('--------------------------------------')
print(f"Barang\t\tHarga\tJumlah\tTotal")
print('--------------------------------------')
print(f"{belanjaan[0]}\t\t{harga_jumlah[0]}\t{harga_jumlah[1]}\t{total_apel}")
print(f"{belanjaan[1]}\t\t{harga_jumlah[2]}\t{harga_jumlah[3]}\t{total_susu}")
print(f"{belanjaan[2]}\t\t{harga_jumlah[4]}\t{harga_jumlah[5]}\t{total_roti}")
print('--------------------------------------')
print(f"Total Belanja\t\t\t{total_belanja}")
print('======================================')
print("    Terima kasih sudah belanja!🙏  ")
print('======================================')
```

```python
# 4. Tambahkan Barang Baru ke List
# Tambahkan "Pisang" ke dalam list belanjaan
belanjaan.append("Pisang")
# Tampilkan kembali isi list belanjaan setelah penambahan
print("Isi list belanjaan setelah ditambahkan:", belanjaan)
```
