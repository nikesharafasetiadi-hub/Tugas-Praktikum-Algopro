Praktikum: Toko Dorayaki Doraemon

```python
#1. Meminta masukan dari user berupa nama makanan, nama karakter, dan kata sifat.
nama_makanan = input("nama makanan:")
nama_karakter = input("nama karakter:")
kata_sifat = input("kata sifat:")
```

```python
#2. Menggabungkan ketiga variabel menjadi satu string nama toko.
nama_toko = nama_makanan + " " + nama_karakter + " " + kata_sifat
print("nama toko:", nama_toko)
```

```python
#3. Mengumpulkan dan menghitung total modal.
modal_doraemon = 555000
modal_nobita = 333000
modal_shizuka = 222000
total_modal = modal_doraemon + modal_nobita + modal_shizuka
print("total modal yang terkumpul: Rp", total_modal)
```

```python
#4. Membuat variabel dan menghitung total pendapatan pada bulan pertama.
jumlah_dorayaki = 350
harga_dorayaki = 5000
total_pendapatan = jumlah_dorayaki * harga_dorayaki
print("total pendapatan bulan pertama: Rp", total_pendapatan)
```

```python
#5. Menghitung keuntungan pada bulan pertama.
keuntungan = total_pendapatan - total_modal
```

```python
#6. Mengidentifikasi jenis error dan menuliskan perbaikan kode yang benar.
#penyebab error tidak ada tanda koma atau tanda tambah yang memisahkan string dan variabel.
print("keuntungan total adalah:", keuntungan)
```
