```python
# a. Membuat Variabel Global
skor_nobita = 0
```

```python
# b. Membuat Fungsi untuk Memperbarui Skor
def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai
```

```python
# c. Membuat Fungsi Aktivitas Nobita

# 1. Fungsi Belajar
def belajar(jam_belajar):
    if jam_belajar < 1:
        update_skor(50)
        output = "Nobita hanya membuka buku sebentar"
    elif 1 <= jam_belajar < 3:
        update_skor(25)
        output = "Nobita belajar secukupnya"
    else:
        update_skor(10)
        output = "Nobita belajar dengan rajin hari ini"
    print(f"Output: {output}")

# 2. Fungsi Tidur
def tidur(jam_tidur):
    if jam_tidur < 1:
        update_skor(-10)
        output = "Nobita kurang tidur"
    elif 1 <= jam_tidur <= 2:
        update_skor(20)
        output = "Waktu tidur Nobita seimbang"
    else:
        update_skor(-20)
        output = "Nobita tidur terlalu lama"
    print(f"Output: {output}")

# 3. Fungsi Main Bersama Doraemon
def main_bersama_Doraemon(waktu_main, jumlah_gadget):
    if waktu_main <= 1 and jumlah_gadget <= 2:
        update_skor(15)
        output = "Nobita bermain secukupnya"
    elif waktu_main <= 2 and jumlah_gadget <= 3:
        update_skor(5)
        output = "Nobita masih bisa mengatur waktu dengan baik"
    else:
        update_skor(-30)
        output = "Nobita kebanyakan bermain"
    print(f"Output: {output}")

# 4. Fugsi Menyontek
def menyontek():
    update_skor(-50)
    print("Ouput: Nobita ketahuan menyontek")
```

```python
# d. Membuat Fungsi Pengendali Aktivitas
def jalankan_aktivitas(pilihan):
    if pilihan == 1:
        jam_b = float(input("Berapa jam Nobita belajar hari ini? "))
        belajar(jam_b)
    elif pilihan == 2:
        jam_t = float(input("Berapa jam Nobita tidur siang hari ini? "))
        tidur(jam_t)
    elif pilihan == 3:
        waktu_main = float(input("Berapa jam Nobita bermain gadget? "))
        jumlah_gadget = int(input("Berapa gadget yang digunakan Nobita? "))
        main_bersama_Doraemon(waktu_main, jumlah_gadget)
    elif pilihan == 4:
        menyontek()
    else:
        print("Pilihan tidak valid, aktivitas hanya ada 4 pilihan")
```

```python
# e. Membuat Fungsi Evaluasi Akhir
def evaluasi_akhir():
    global skor_nobita
    if skor_nobita >= 80:
        output = "Siswa Teladan"
    elif 50 <= skor_nobita < 80:
        output = "Cukup Rajin"
    else:
        output = "Pemalas"
    
    print("\n" + "="*53)
    print("=========== EVALUASI AKHIR PERILAKU NOBITA ==========")
    print(f"Total Skor Akhir : {skor_nobita}")
    print(f"Status Perilaku  : {output}")
    print("="*53)
```

```python
# f. Membuat Menu Utama Program Peningkatan Disiplin Harian

def main():
    while True:
        print("\n========= PROGRAM PENINGKAT DISIPLIN NOBITA =========")
        print("1. Belajar")
        print("2. Tidur")
        print("3. Main bersama Doraemon")
        print("4. Menyontek saat ujian")
        print("5. Selesai & Lihat Evaluasi")
        
        try:
            pilih = int(input("Pilih aktivitas (1-5): "))
            if pilih == 5:
                evaluasi_akhir()
                print("Program selesai. Terima kasih!")
                break
            jalankan_aktivitas(pilih)
            print(f"Skor Nobita saat ini: {skor_nobita}")
        except ValueError:
            print("Masukkan angka yang valid!")

# Menjalankan program utama
if __name__ == "__main__":
    main()
```
